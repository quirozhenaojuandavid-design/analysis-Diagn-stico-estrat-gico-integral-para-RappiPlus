## 🔹 Paso 2: Analizar si el negocio es rentable

### 2.1 Cálculo de KPIs principales

**🎯 Objetivo:** Calcular los indicadores clave del negocio para evaluar ingresos, costos y rentabilidad.

Se usan los 3 datasets (`orders`, `catalog`, `marketing_spend`):

**📊 Parte 1: Rentabilidad del negocio**
- ¿Cuál es el ingreso total (revenue)? 
- ¿Cuál es el costo total? 
- ¿Cuánto se ha invertido en marketing? 
- ¿El negocio es rentable? (calcular profit)  

---

**📈 Parte 2: Comportamiento de ventas**
- ¿Cuál es el ticket promedio por orden? 
- ¿Cuál es la cantidad promedio de productos por orden? 
- ¿Cuál es el producto más vendido?
- ¿Cuánto se ha gastado en marketing por canal? 

```python
# ==========================================
# PASO 2.1 - KPIs PRINCIPALES
# ==========================================

# 1. Revenue total
revenue_total = orders_clean['monto_total'].sum()

# 2. Unir pedidos con catálogo para obtener costos
orders_costos = orders_clean.merge(
    catalog_clean[['nombre_producto', 'costo_unitario']],
    on='nombre_producto',
    how='left'
)

# Verificar que todos los pedidos tengan costo
print(
    "Pedidos sin costo asociado:",
    orders_costos['costo_unitario'].isna().sum()
)

# 3. Costo por pedido
orders_costos['costo_total'] = (
    orders_costos['cantidad'] *
    orders_costos['costo_unitario']
)

# 4. Costo total
costo_total = orders_costos['costo_total'].sum()

# 5. Marketing total
marketing_total = marketing_clean['gasto'].sum()

# 6. Profit
profit = revenue_total - costo_total - marketing_total

# 7. Margen
margen = profit / revenue_total * 100

# 8. Ticket promedio
ticket_promedio = orders_clean['monto_total'].mean()

# 9. Cantidad promedio
cantidad_promedio = orders_clean['cantidad'].mean()

# 10. Producto más vendido
productos_vendidos = (
    orders_clean
    .groupby('nombre_producto')['cantidad']
    .sum()
    .sort_values(ascending=False)
)

producto_mas_vendido = productos_vendidos.idxmax()
cantidad_mas_vendida = productos_vendidos.max()

# 11. Marketing por canal
marketing_por_canal = (
    marketing_clean
    .groupby('canal')['gasto']
    .sum()
    .sort_values(ascending=False)
)

# ==========================================
# RESULTADOS
# ==========================================

print("\n===== KPIs DEL NEGOCIO =====")

print(f"Revenue total: ${revenue_total:,.2f}")
print(f"Costo total: ${costo_total:,.2f}")
print(f"Marketing total: ${marketing_total:,.2f}")
print(f"Profit: ${profit:,.2f}")
print(f"Margen de rentabilidad: {margen:.2f}%")

print("\n===== COMPORTAMIENTO DE VENTAS =====")

print(f"Ticket promedio: ${ticket_promedio:,.2f}")
print(f"Productos promedio por pedido: {cantidad_promedio:.2f}")
print(f"Producto más vendido: {producto_mas_vendido}")
print(f"Unidades vendidas: {cantidad_mas_vendida:,.0f}")

print("\n===== MARKETING POR CANAL =====")
display(marketing_por_canal)
```

```python
# PARTE 2: Conversiones
# ======================

query_conversion = '''

WITH funnel AS (
    SELECT
        CASE
            WHEN nombre_evento = 'first_visit' THEN 1
            WHEN nombre_evento = 'add_to_cart' THEN 2
            WHEN nombre_evento = 'select_item' THEN 3
            WHEN nombre_evento = 'begin_checkout' THEN 4
            WHEN nombre_evento = 'add_payment_info' THEN 5
            WHEN nombre_evento = 'purchase' THEN 6
        END AS etapa,
        
        nombre_evento,
        COUNT(DISTINCT id_usuario) AS usuarios
    FROM events
    WHERE nombre_evento IN (
        'first_visit',
        'add_to_cart',
        'select_item',
        'begin_checkout',
        'add_payment_info',
        'purchase'
    )
    GROUP BY nombre_evento
),

conversion AS (
    SELECT
        etapa,
        nombre_evento,
        usuarios,
        LAG(usuarios) OVER (
            ORDER BY etapa
        ) AS usuarios_etapa_anterior
    FROM funnel
)

SELECT
    etapa,
    nombre_evento,
    usuarios,
    usuarios_etapa_anterior,
    ROUND(
        usuarios * 100.0 / usuarios_etapa_anterior,
        2
    ) AS tasa_conversion,
    ROUND(
        100.0 - (usuarios * 100.0 / usuarios_etapa_anterior),
        2
    ) AS drop_off
FROM conversion
ORDER BY etapa;
'''

conversion = pd.read_sql(query_conversion, con=engine)
conversion
```

## 🔹 Paso 4: Evaluar si los usuarios regresan (retención por cohortes)

**🎯 Objetivo:** Analizar la retención de usuarios para entender si regresan después de registrarse.

**Tablas**

- `users` 
- `user_activity` 

---
1. Se identifica la cohorte de cada usuario según el **mes de registro**.


2. Se calcula la retención semanal: cuántos usuarios **se mantienen activos** en cada semana desde su registro.
   - `retenido_w1`: usuarios activos en la semana 1  
   - `retenido_w2`: usuarios activos en la semana 2  
   - `retenido_w3`: usuarios activos en la semana 3  


3. Se calcula el porcentaje de retención para cada semana, dividiendo los usuarios retenidos entre los clientes iniciales de la cohorte:  
   - `semana_1`: porcentaje de usuarios retenidos en la semana 1  
   - `semana_2`: porcentaje de usuarios retenidos en la semana 2  
   - `semana_3`: porcentaje de usuarios retenidos en la semana 3  

Se revisa que la columna de fecha esté en formato correcto (`DATE`).  
Se realiza la conversión usando: `CAST(fecha_registro AS DATE)`

```python
# Explorar tabla users
# =========================
query_users = '''
SELECT *
FROM users;
'''
users = pd.read_sql(query_users, con=engine)
users.head(3)
```

```python
# Explorar tabla user_activity
# =========================
query_user_activity = '''
SELECT *
FROM user_activity;
'''

user_activity = pd.read_sql(
    query_user_activity,
    con=engine
)

user_activity.head(3)
```

```python
# Retención por cohortes
# ======================

query_cohort_retention_final = '''

WITH cohortes AS (
    SELECT
        id_usuario,
        DATE_TRUNC('month', CAST(fecha_registro AS DATE)) AS cohorte
    FROM users
),

actividad_semanal AS (
    SELECT
        id_usuario,
        MAX(
            CASE
                WHEN dias_despues_registro BETWEEN 1 AND 7
                     AND activo = 1
                THEN 1
                ELSE 0
            END
        ) AS retenido_w1,

        MAX(
            CASE
                WHEN dias_despues_registro BETWEEN 8 AND 14
                     AND activo = 1
                THEN 1
                ELSE 0
            END
        ) AS retenido_w2,

        MAX(
            CASE
                WHEN dias_despues_registro BETWEEN 15 AND 21
                     AND activo = 1
                THEN 1
                ELSE 0
            END
        ) AS retenido_w3

    FROM user_activity
    GROUP BY id_usuario
),

cohortes_final AS (
    SELECT
        c.cohorte,

        COUNT(DISTINCT c.id_usuario) AS usuarios_iniciales,

        COUNT(
            DISTINCT CASE
                WHEN a.retenido_w1 = 1
                THEN c.id_usuario
            END
        ) AS retenido_w1,

        COUNT(
            DISTINCT CASE
                WHEN a.retenido_w2 = 1
                THEN c.id_usuario
            END
        ) AS retenido_w2,

        COUNT(
            DISTINCT CASE
                WHEN a.retenido_w3 = 1
                THEN c.id_usuario
            END
        ) AS retenido_w3

    FROM cohortes c
    LEFT JOIN actividad_semanal a
        ON c.id_usuario = a.id_usuario

    GROUP BY c.cohorte
)

SELECT
    cohorte,
    usuarios_iniciales,
    retenido_w1,
    retenido_w2,
    retenido_w3,

    ROUND(
        retenido_w1 * 100.0 / NULLIF(usuarios_iniciales, 0),
        2
    ) AS semana_1,

    ROUND(
        retenido_w2 * 100.0 / NULLIF(usuarios_iniciales, 0),
        2
    ) AS semana_2,

    ROUND(
        retenido_w3 * 100.0 / NULLIF(usuarios_iniciales, 0),
        2
    ) AS semana_3

FROM cohortes_final
ORDER BY cohorte;
'''

cohorte_final = pd.read_sql(
    query_cohort_retention_final,
    con=engine
)

cohorte_final

# Ejecutar la consulta
cohorte_final = pd.read_sql(query_cohort_retention_final, con=engine)
cohorte_final
```

## 🔹 Paso 5: Validar si los cambios generan impacto (test estadístico)

🎯 **Objetivo:** Evaluar si la modificación en la UI del checkout impacta la **tasa de conversión de compra**.

---

1. **Analizar el dataset** `experiment_checkout_ui.csv` para identificar la métrica principal **conversion**.
   - La métrica **conversion** es 1 si el usuario completó la compra, 0 si no.    
2. **Plantear la hipótesis estadística**     
3. **Aplicar el test estadístico adecuado** 
4. **Interpretar el resultado**  

---
Hipótesis estadística
   - **H₀ (Hipótesis nula):** ...
   - **H₁ (Hipótesis alternativa):** ...
   
**Test estadístico:** ...  
**Nivel de significancia alpha:** ...

```python
# ==========================================
# PASO 5 - CARGAR DATASET DEL EXPERIMENTO
# ==========================================

import pandas as pd
import numpy as np

url_experiment = 'https://practicum-content.s3.amazonaws.com/datasets/experiment_checkout_ui.csv'

experiment = pd.read_csv(url_experiment)

print("Dataset cargado correctamente.")
print("Dimensiones:", experiment.shape)

print("\nColumnas del dataset:")
print(experiment.columns.tolist())

display(experiment.head())
```

```python
print(experiment.columns.tolist())
```

```python
experiment['variante'].value_counts()
```

```python

# ==========================================
# 2. REVISAR GRUPOS DEL EXPERIMENTO
# ==========================================

print("Usuarios por variante:")
print(experiment['variante'].value_counts())

```

```python
# ==========================================
# DIAGNÓSTICO DEL DATASET
# ==========================================

print("Columnas actuales de experiment:")
print(experiment.columns.tolist())

print("\nPrimeras filas:")
display(experiment.head())
```

```python
# ==========================================
# 3. REVISAR VARIABLE DE CONVERSIÓN
# ==========================================

print("Valores de convirtio:")
print(experiment['convirtio'].value_counts())

print("\nTipo de dato:")
print(experiment['convirtio'].dtype)
```

```python
# ==========================================
# 4. TASA DE CONVERSIÓN POR VARIANTE
# ==========================================

conversion_control = experiment.loc[
    experiment['variante'] == 'control',
    'convirtio'
].mean()

conversion_tratamiento = experiment.loc[
    experiment['variante'] == 'tratamiento',
    'convirtio'
].mean()

print("===== TASA DE CONVERSIÓN =====")

print(f"Control: {conversion_control * 100:.2f}%")
print(f"Tratamiento: {conversion_tratamiento * 100:.2f}%")
```

```python
# ==========================================
# 5. DIFERENCIA ENTRE VARIANTES
# ==========================================

diferencia = (
    conversion_tratamiento - conversion_control
) * 100

print(
    f"Diferencia: {diferencia:.2f} puntos porcentuales"
)
```

```python

# ==========================================
# 6. PREPARAR DATOS PARA EL TEST ESTADÍSTICO
# ==========================================

# Número de conversiones en cada grupo
conversiones_control = experiment.loc[
    experiment['variante'] == 'control',
    'convirtio'
].sum()

conversiones_tratamiento = experiment.loc[
    experiment['variante'] == 'tratamiento',
    'convirtio'
].sum()

# Número total de usuarios en cada grupo
usuarios_control = (
    experiment['variante'] == 'control'
).sum()

usuarios_tratamiento = (
    experiment['variante'] == 'tratamiento'
).sum()

# Mostrar resultados
print("===== DATOS DEL EXPERIMENTO =====")

print(
    f"Control: {conversiones_control} conversiones "
    f"de {usuarios_control} usuarios"
)

print(
    f"Tratamiento: {conversiones_tratamiento} conversiones "
    f"de {usuarios_tratamiento} usuarios"
)
```

```python
from statsmodels.stats.proportion import proportions_ztest

conversiones = [
    conversiones_control,
    conversiones_tratamiento
]

usuarios = [
    usuarios_control,
    usuarios_tratamiento
]

stat, p_value = proportions_ztest(
    count=conversiones,
    nobs=usuarios,
    alternative='two-sided'
)

alpha = 0.05

print("===== RESULTADO DEL TEST =====")
print(f"Estadístico z: {stat:.4f}")
print(f"p-value: {p_value:.4f}")
print(f"Nivel de significancia α: {alpha}")
```

```python
# ==========================================
# 8. DECISIÓN ESTADÍSTICA
# ==========================================

if p_value < alpha:
    print("RECHAZAMOS H₀")
    print(
        "Existe una diferencia estadísticamente significativa "
        "entre las tasas de conversión."
    )
else:
    print("NO RECHAZAMOS H₀")
    print(
        "No existe evidencia suficiente para afirmar que "
        "las tasas de conversión sean diferentes."
    )
```
